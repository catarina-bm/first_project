```python
# gives the different certificate with count, including NaN
df_clean["certificate"].value_counts(dropna=False)  
```

```python
# Replace NaN values with Unrated
df_clean["certificate"]= df_clean["certificate"].fillna("Unrated")
```

```python
# ratings from India, US and TV are grouped into 4 audience levels. 

age_groups = {"U": "All ages",
              "G": "All ages",
              "Passed": "All ages",
              "Approved": "All ages",
              "UA": "Family/Teen",
              "U/A": "Family/Teen",
              "PG-13": "Family/Teen",
              "PG": "Family/Teen",
              "GP": "Family/Teen",
              "TV-PG": "Family/Teen",
              "TV-14": "Family/Teen",
              "A": "Adult",
              "R": "Adult",
              "TV-MA": "Adult",
              "16": "Adult",
              "Unrated": "Unrated"    
    }
df_clean["audience"] = df_clean["certificate"].map(age_groups)
```

```python
df_clean["gross_revenue"] = (
    df_clean["gross_revenue"]
    .astype(str)
    .str.replace(",", "", regex=False)
)
df_clean["gross_revenue"] = pd.to_numeric(df_clean["gross_revenue"], errors="coerce").round().astype("Int64")
```
