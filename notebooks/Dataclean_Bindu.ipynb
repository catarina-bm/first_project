First_project_day1_Bindu: Deal with the missing values 

```python
import pandas as pd
```

```python
url = r"../data/raw/imdb_top_1000.csv"
df_clean = pd.read_csv(url)
df_clean.head()
```

```python
df_clean.columns = df_clean.columns.str.strip().str.lower().str.replace(" ", "_")
```

```python
df_clean = df_clean.rename(columns={"series_title": "movie_title",
                        "no_of_votes": "number_of_votes",
                        "gross": "gross_revenue"})
```

```python
#Missing values
df_clean.isna().sum()
# 100 out of 1000 certificate values are null
```

```python
#to know data types of each column
df_clean.dtypes
```

```python
# gives the different certificate with count, including NaN
df_clean["certificate"].value_counts(dropna=False)  
```

```python
# Replace NaN values with Unrated
df_clean["certificate"]= df_clean["certificate"].fillna("Unrated")
```

```python
# ratings from India, US and TV are grouped into 4 audience levels. 

age_groups = {"U": "All ages",
              "G": "All ages",
              "Passed": "All ages",
              "Approved": "All ages",
              "UA": "Family/Teen",
              "U/A": "Family/Teen",
              "PG-13": "Family/Teen",
              "PG": "Family/Teen",
              "GP": "Family/Teen",
              "TV-PG": "Family/Teen",
              "TV-14": "Family/Teen",
              "A": "Adult",
              "R": "Adult",
              "TV-MA": "Adult",
              "16": "Adult",
              "Unrated": "Unrated"    
    }
df_clean["audience"] = df_clean["certificate"].map(age_groups)
display(df_clean["audience"])
```

```python
df_clean["audience"].value_counts() # ratings from India, US and TV are grouped into 4 audience levels. 
```

```python
df_clean["audience"].isna().sum() # no NaN values
```

```python
df_clean["meta_score"].unique() 
```

```python
df_clean["Main_genre"]=df_clean["genre"].str.split(",").str[0].str.strip()
df_clean["Main_genre"].unique().tolist()
```

```python
#checking the missing values with the Main_genre "Fantasy"
df_clean[(df_clean["meta_score"].isna()) & (df_clean["Main_Genre"]== "Fantasy")]
```

```python
# Filling the remaining NaN values with the overall median
df_clean["meta_score"] = df_clean["meta_score"].fillna(df_clean["meta_score"].median())
```

```python
df_clean["meta_score"].isna().sum()
```

```python
df_clean["gross_revenue"].isna().sum() ##169 films have no US box office data and were left out of revenue analysis. For popularity among all 1000 films, use No_of_Votes.
```

```python
# split "Crime, Drama" into separate columns (max 3 genres per film)
genre_split = df_clean["genre"].str.split(",", expand=True)
genre_split.columns = ["genre_1", "genre_2", "genre_3"]

# remove the space left after each comma
for col in genre_split.columns:
    genre_split[col] = genre_split[col].str.strip()

df_clean = df_clean.drop(columns=["genre_1", "genre_2", "genre_3"], errors="ignore")
df_clean = pd.concat([df_clean, genre_split], axis=1)
#df_clean[["genre", "genre_1", "genre_2", "genre_3"]].head()
display(df_clean)
```

```python
df_clean["gross_revenue"] = (df_clean["gross_revenue"]
                     .str.strip()
                     .str.replace(",", "", regex=False))
df_clean["gross_revenue"] = pd.to_numeric(df_clean["gross_revenue"], errors="coerce")

#  millions, easier to read in charts
df_clean["gross_revenue"] = df_clean["gross_revenue"] / 1_000_000
```
